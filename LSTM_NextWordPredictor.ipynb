```python
def text_to_index(text,vocab):
  text_index=[]
  for token in text:
    if token in vocab:
      text_index.append(vocab[token])
    else:
      text_index.append(vocab['<UNK>'])
  return text_index
```

```python
input_sentences_total=[]
for lines in splited_lines:
  input_sentences_total.append(text_to_index(word_tokenize(lines.lower()),vocab))
```

```python
len(input_sentences_total)
```

```python
training_seq=[]
for lines in input_sentences_total:
  for i in range(len(lines)):
    training_seq.append(lines[:i+1])
```

```python
len_list=[]
for length in training_seq:
  len_list.append(len(length))
max(len_list)
```

```python
padded_training_seq=[]
for seq in training_seq:
  padded_training_seq.append(([0]*(max(len_list)-len(seq)))+seq)
```

```python
padded_training_seq=torch.tensor(padded_training_seq, dtype=torch.long)
```

```python
x=padded_training_seq[:,:-1]
y=padded_training_seq[:,-1]
```

```python
class CustomDataset(Dataset):
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def __len__(self):
    return self.x.shape[0]
  def __getitem__(self,index):
    return self.x[index],self.y[index]
```

```python
dataset=CustomDataset(x,y)
```

```python
dataloader=DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
class MyModel(nn.Module):
  def __init__(self,vocab_size):
    super().__init__()
    self.embedding=nn.Embedding(vocab_size,100)
    self.lstm=nn.LSTM(100,150,batch_first=True)
    self.fc=nn.Linear(150,vocab_size)
  def forward(self,question):
    embeddings=self.embedding(question)
    inter_hidden_state,(final_hidden_state,final_cell_state)=self.lstm(embeddings)
    output=self.fc(final_hidden_state.squeeze(0))
    return output
```

```python
model=MyModel(len(vocab))
```

```python
model.to(device)
```

```python
epochs=50
learning_rate=0.01
criterian=nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate, weight_decay=1e-5)
```

```python
for epoch in range(epochs):
  total_loss=0
  for batch_x,batch_y in dataloader:
    batch_x=batch_x.to(device)
    batch_y=batch_y.to(device)
    optimizer.zero_grad()
    output=model(batch_x)
    loss=criterian(output,batch_y)
    loss.backward()
    optimizer.step()
    total_loss=total_loss+loss.item()
  print(f'epoch {epoch+1} loss {total_loss}')
```

```python
# Function to calculate accuracy
def calculate_accuracy(model, dataloader, device):
    model.eval()  # Set the model to evaluation mode
    correct = 0
    total = 0

    with torch.no_grad():  # No need to compute gradients
        for batch_x, batch_y in dataloader1:
            batch_x, batch_y = batch_x.to(device), batch_y.to(device)

            # Get model predictions
            outputs = model(batch_x)

            # Get the predicted word indices
            _, predicted = torch.max(outputs, dim=1)

            # Compare with actual labels
            correct += (predicted == batch_y).sum().item()
            total += batch_y.size(0)

    accuracy = correct / total * 100
    return accuracy

# Compute accuracy
accuracy = calculate_accuracy(model, dataloader, device)
print(f"Model Accuracy: {accuracy:.2f}%")
```
